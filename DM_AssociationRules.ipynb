```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules

# 1. ساخت 20 تراکنش مسیر گل برای تیم پیروزی
transactions = [
    ["توپ‌گیری", "پاس کوتاه", "پاس در عمق", "شوت"],
    ["ضربه کاشته", "پاس عرضی", "شوت"],
    ["توپ‌گیری", "دریبل", "پاس در عمق", "شوت"],
    ["اوت دستی", "پاس بلند", "شوت"],
    ["توپ‌گیری", "پاس کوتاه", "پاس کوتاه", "شوت"],
    ["ضربه کاشته", "پاس در عمق", "شوت"],
    ["پاس عرضی", "پاس در عمق", "شوت"],
    ["توپ‌گیری", "پاس کوتاه", "پاس در عمق", "شوت"],
    ["دریبل", "پاس در عمق", "شوت"],
    ["توپ‌گیری", "پاس در عمق", "شوت"],
    ["پاس عرضی", "پاس در عمق", "شوت"],
    ["ضربه کاشته", "شوت"],
    ["توپ‌گیری", "دریبل", "شوت"],
    ["توپ‌گیری", "پاس کوتاه", "شوت"],
    ["اوت دستی", "پاس بلند", "پاس در عمق", "شوت"],
    ["ضربه کاشته", "پاس کوتاه", "پاس در عمق", "شوت"],
    ["توپ‌گیری", "پاس در عمق", "شوت"],
    ["دریبل", "پاس عرضی", "شوت"],
    ["توپ‌گیری", "پاس بلند", "شوت"],
    ["توپ‌گیری", "پاس در عمق", "شوت"]
]

# 2. تبدیل داده‌ها به فرمت مناسب
te = TransactionEncoder()
te_data = te.fit(transactions).transform(transactions)
df = pd.DataFrame(te_data, columns=te.columns_)

# 3. کشف آیتم‌ست‌های پرتکرار
frequent_itemsets = apriori(df, min_support=0.2, use_colnames=True)

# 4. استخراج قواعد انجمنی
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.6)

# 5. مرتب‌سازی بر اساس lift
rules = rules.sort_values(by="lift", ascending=False)
```

```python

# 6. نمایش خلاصه خروجی
rules_cleaned = rules.copy()
rules_cleaned["antecedents"] = rules_cleaned["antecedents"].apply(lambda x: ', '.join(x))
rules_cleaned["consequents"] = rules_cleaned["consequents"].apply(lambda x: ', '.join(x))

for index, row in rules_cleaned.iterrows():
    print(f"{row['antecedents']:<20} -> {row['consequents']:<20} | support: {row['support']:.2f} | confidence: {row['confidence']:.2f} | lift: {row['lift']:.2f}")
```
